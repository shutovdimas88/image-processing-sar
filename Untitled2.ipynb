```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import mean_squared_error as mse_func

print("✅ Импорты выполнены")
print("cv2", cv2.__version__)
print("numpy", np.__version__)
```

```python
import os

# Проверяем, какие файлы есть
files = os.listdir('.')
print("Файлы в папке:")
for f in files:
    if not f.endswith('.ipynb'):    # пропускаем ноутбуки
        print("  •", f)

img = cv2.imread('sar_1_gray.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("\n❌ Не удалось прочитать sar_1_gray.jpg")
else:
    print("\n✅ Изображение загружено")
    print("shape =", img.shape)
    print("dtype =", img.dtype)
    print("min =", img.min(), " max =", img.max())
    print("mean =", round(float(img.mean()), 2), " std =", round(float(img.std()), 2))
```

```python
plt.figure(figsize=(10, 6))
plt.imshow(img, cmap='gray')
plt.title('sar_1_gray.jpg')
plt.axis('off')
plt.show()
```

```python
hist      = cv2.calcHist([img], [0], None, [256], [0, 256])
hist_cum  = hist.cumsum()
hist_norm = hist / img.size

fig, axes = plt.subplots(1, 3, figsize=(18, 4))
axes[0].plot(hist, color='black');       axes[0].set_title('Абсолютная H(i)')
axes[1].plot(hist_cum, color='blue');    axes[1].set_title("Кумулятивная H'(i)")
axes[2].plot(hist_norm, color='green');  axes[2].set_title('Нормированная')
for ax in axes:
    ax.set_xlabel('Яркость'); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()

print("Пик яркости (mode):", int(hist.argmax()))
print("Медиана яркости:", int(np.argmax(hist_cum >= img.size/2)))
print("Сумма пикселей в гистограмме:", int(hist.sum()), "= img.size:", img.size)
```

```python
def gamma_correction(image, gamma):
    norm = image.astype(np.float32) / 255.0
    return np.clip((norm ** gamma) * 255.0, 0, 255).astype(np.uint8)

img_gamma_low  = gamma_correction(img, 0.4)   # γ < 1 → светлее
img_gamma_high = gamma_correction(img, 2.5)   # γ > 1 → темнее

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img, cmap='gray');             axes[0].set_title('γ = 1 (исходное)')
axes[1].imshow(img_gamma_low, cmap='gray');   axes[1].set_title('γ = 0.4 (<1)')
axes[2].imshow(img_gamma_high, cmap='gray');  axes[2].set_title('γ = 2.5 (>1)')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()

print("mean γ=1.0:", round(float(img.mean()), 2))
print("mean γ=0.4:", round(float(img_gamma_low.mean()), 2))
print("mean γ=2.5:", round(float(img_gamma_high.mean()), 2))
```

```python
def compare(a, b, label):
    m = mse_func(a, b)
    s = ssim(a, b, data_range=255)
    print(f"{label:<30} | MSE = {m:10.2f} | SSIM = {s:.4f}")

print("=== Сравнение ===")
compare(img, img_gamma_low,  "Original vs γ=0.4")
compare(img, img_gamma_high, "Original vs γ=2.5")
```

```python
# Эталон eq_gray: если файла нет — берём equalizeHist(img)
try:
    eq_gray = cv2.imread('eq_gray.jpg', cv2.IMREAD_GRAYSCALE)
    if eq_gray is None:
        raise FileNotFoundError
    print("Использую eq_gray.jpg")
except FileNotFoundError:
    eq_gray = cv2.equalizeHist(img)
    print("eq_gray.jpg не найден — использую equalizeHist(img)")

def stat_correction(src, ref):
    E_t, D_t = float(src.mean()), float(src.std())
    E_s, D_s = float(ref.mean()), float(ref.std())
    if D_t == 0: D_t = 1e-6
    return np.clip(E_s + (src.astype(np.float32) - E_t) * (D_s / D_t),
                   0, 255).astype(np.uint8)

img_stat = stat_correction(img, eq_gray)
compare(img, img_stat, "Original vs stat_corrected")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img, cmap='gray');      axes[0].set_title('Исходное')
axes[1].imshow(eq_gray, cmap='gray');  axes[1].set_title('Эталон eq_gray')
axes[2].imshow(img_stat, cmap='gray'); axes[2].set_title('После стат. коррекции')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()

print("Исходное:  E_t =", round(float(img.mean()),2), " D_t =", round(float(img.std()),2))
print("Эталон:    E_s =", round(float(eq_gray.mean()),2), " D_s =", round(float(eq_gray.std()),2))
print("Результат: E   =", round(float(img_stat.mean()),2), " D   =", round(float(img_stat.std()),2))
```

```python
print("=== Пороговая фильтрация ===")
bins_fixed = {}
for t in [50, 100, 150, 200]:
    _, b = cv2.threshold(img, t, 255, cv2.THRESH_BINARY)
    bins_fixed[t] = b
    print(f"THRESH_BINARY T={t:<3} → доля белых = {np.count_nonzero(b)/b.size:.4f}")

t_otsu, bin_otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print(f"OTSU → T* = {t_otsu}")

adap_mean  = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                   cv2.THRESH_BINARY, 21, 5)
adap_gauss = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                   cv2.THRESH_BINARY, 21, 5)
bin_inv    = cv2.threshold(img, t_otsu, 255, cv2.THRESH_BINARY_INV)[1]

fig, axes = plt.subplots(2, 4, figsize=(18, 9))
imgs   = [img, bins_fixed[50], bins_fixed[100], bins_fixed[150],
          bin_otsu, bin_inv, adap_mean, adap_gauss]
titles = ['Исходное', 'T=50', 'T=100', 'T=150',
          f'OTSU (T={t_otsu})', 'BINARY_INV', 'Adaptive mean', 'Adaptive gauss']
for ax, im, ttl in zip(axes.ravel(), imgs, titles):
    ax.imshow(im, cmap='gray'); ax.set_title(ttl); ax.axis('off')
plt.tight_layout(); plt.show()
```
